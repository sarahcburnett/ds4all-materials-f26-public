## Section 1: Building a tibble

A tibble is built from a label and a vector for each column.

```r
tibble(`Number of petals` = c(8, 34, 5))
```

Note the backticks around `` `Number of petals` ``. We need those any time a
column name has a space or another special character in it.

Two columns, or more, work the same way.

```r
flowers <- tibble(
  `Number of petals` = c(8, 34, 5),
  Name = c("lotus", "sunflower", "rose")
)

flowers
```

All columns have to be the same length. Uncomment the next line and run it to
see the error.

```r
# tibble(x = c(1, 2, 3), y = c("a", "b"))
```

`mutate()` adds a column. It returns a **new** tibble and leaves the original
alone.

```r
flowers |>
  mutate(Color = c("pink", "yellow", "red"))
```

```r
# flowers itself has not changed
flowers
```

To make the change stick, we assign it back.

```r
flowers <- flowers |>
  mutate(Color = c("pink", "yellow", "red"))

flowers
```

## Section 2: Getting things out of a tibble

Three operations that look similar and give back different things.

`pull()` gives the bare vector of values in a column.

```r
flowers |> pull(Name)
```

`select()` gives back a tibble, even when we ask for a single column.

```r
flowers |> select(Name)
```

Compare the two outputs above. One is a vector. One is a table that happens to
have one column. That difference matters when we pass the result to another
function.

We can also ask for a column by position. `Name` is the second column.

```r
flowers |> pull(2)
```

Once we have a vector, square brackets get us a single element.

**In R, positions start at 1.** The first element is `[1]`, not `[0]`.

```r
petals <- flowers |> pull(`Number of petals`)

petals
petals[1]
petals[3]
```

## Section 3: Functions on columns

A tibble column is a vector, so the two kinds of function we met before both
work on columns with no change.

**Aggregate** functions collapse a vector to a single number.

```r
mean(petals)
sum(petals)
max(petals)
length(petals)
```

**Elementwise** functions return a vector the same length as the input.

```r
sqrt(petals)
round(sqrt(petals), 2)
```

An aggregate result is usually something we just look at. An elementwise result
is usually something we want to keep, and `mutate()` is how we keep it.

```r
flowers |>
  mutate(`Petals per flower, scaled` = petals / max(petals))
```

Notice that the expression inside `mutate()` used both kinds at once: `petals`
elementwise, divided by `max(petals)`, an aggregate. R recycles the single
number across the whole column.

We can also refer to columns by name inside `mutate()`, without pulling them
out first. This is the form we will use from here on.

```r
flowers |>
  mutate(Scaled = `Number of petals` / max(`Number of petals`))
```

## Section 5: Importing a CSV

`read_csv()` comes from `readr`, part of the core tidyverse.

The path is relative to where this notebook lives. A file inside a `data/`
folder sitting next to the notebook is `"data/films.csv"`.

```r
films <- read_csv("data/films.csv", show_col_types = FALSE)
```

`show_col_types = FALSE` turns off the message about how each column was
guessed. Leave it off while you are learning; it is useful, just noisy.

### Look at what you loaded

Do this every time. It is how you catch a column that read in as text when it
should have been a number.

```r
head(films)
```

```r
glimpse(films)
```

In RStudio there is a third option, `View(films)`, which opens the whole table
in a scrollable tab. It does not work in a notebook on the Hub. Here, the
equivalent is to print a slice.

```r
films |> slice(1:20)
```

### Writing one back out

Same shape as reading, with the arguments reversed: the object first, then
where it should go.

```r
write_csv(films, "data/films-clean.csv")
```
